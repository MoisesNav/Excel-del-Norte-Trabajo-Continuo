```python
import requests
import pandas as pd

def generar_df_exel():
    """
    Obtiene datos de productos, fichas técnicas e imágenes desde la API de Exel del Norte y los combina en un DataFrame.
    
    Retorna:
    - df_exel (pd.DataFrame): DataFrame consolidado con productos, fichas técnicas e imágenes.
    """
    base_url = "https://api01.exeldelnorte.com.mx/"
    api_key = "API_6ee6a508bb7ebd80d19ef1d6edf141354cade0f7"
    
    headers = {"Authorization": api_key}
    
    # Obtener datos de productos
    try:
        response = requests.get(base_url + "productos", headers=headers, params={"sin_stock": "false"})
        response.raise_for_status()
        productos = response.json().get("datos", [])
        df_productos = pd.DataFrame(productos) if productos else pd.DataFrame()
    except requests.exceptions.RequestException as e:
        print(f"Error al obtener productos: {e}")
        return pd.DataFrame()
    
    # Obtener ficha técnica
    try:
        response = requests.get(base_url + "productos_fichatecnica", headers=headers)
        response.raise_for_status()
        tecnica = response.json().get("datos", [])
        df_tecnica = pd.DataFrame(tecnica) if tecnica else pd.DataFrame()
    except requests.exceptions.RequestException as e:
        print(f"Error al obtener ficha técnica: {e}")
        df_tecnica = pd.DataFrame()  # Si falla, seguimos sin ficha técnica
    
    # Unir productos con ficha técnica si ambas tablas tienen datos
    if not df_productos.empty and not df_tecnica.empty:
        df_exel = df_productos.merge(df_tecnica[['sku', 'ficha_tecnica']], on='sku', how='left')
    else:
        df_exel = df_productos
    
    # Obtener imágenes
    try:
        response = requests.get(base_url + "imagenes", headers=headers)
        response.raise_for_status()
        imagenes = response.json().get("DATA", [])
        df_imagenes = pd.DataFrame(imagenes) if imagenes else pd.DataFrame()
    except requests.exceptions.RequestException as e:
        print(f"Error al obtener imágenes: {e}")
        df_imagenes = pd.DataFrame()  # Si falla, seguimos sin imágenes

    # Unir imágenes si existen
    if not df_imagenes.empty:
        df_exel = df_exel.merge(df_imagenes[['sku', 'imagenes']], on='sku', how='left')

    return df_exel
```
