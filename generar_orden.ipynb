# Confirmar Orden

```python


# Tu API Key
API_KEY = "API_6ee6a508bb7ebd80d19ef1d6edf141354cade0f7"
url = "https://api01.exeldelnorte.com.mx/confirma_preorden"

# El número de la pre-orden que deseas confirmar
num_orden_a_confirmar = "905782221"

# Parámetros de la URL (Query Parameters)
querystring = {
    "num_orden": num_orden_a_confirmar
}

# Agregar los Headers
headers = {
    'Authorization': API_KEY
}

try:
    # Ejecuta la petición GET usando 'params' para inyectar el número de orden de forma limpia en la URL
    response = requests.request("GET", url, headers=headers, params=querystring)

    print("Status:", response.status_code)
    
    try:
        # Intenta imprimir la respuesta en formato JSON estructurado
        print("Respuesta:", json.dumps(response.json(), indent=4, ensure_ascii=False))
    except json.JSONDecodeError:
        # Si la respuesta no viene en formato JSON válido, imprime el texto plano
        print("Respuesta:", response.text)

except requests.exceptions.RequestException as e:
    print("Error en la petición de red:", e)
```
